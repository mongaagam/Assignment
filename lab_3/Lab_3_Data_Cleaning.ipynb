## 1. Profile Missingness

In this section, we identify missing values in each column.

We calculate both the number of missing values and the percentage of missing values for every column.

```python
import pandas as pd
df = pd.read_csv("/Users/agam/Desktop/pythonAssignment/week_challenge/samples/file1.csv")
df.head()
```

## Missing Value Count

We calculate the total number of missing values in each column.

```python
df.isna().sum()
```

## Missing Value Percentage

We calculate the percentage of missing values in each column to understand how much data is missing.

```python
mis_per = df.isna().mean() * 100
mis_per
```
