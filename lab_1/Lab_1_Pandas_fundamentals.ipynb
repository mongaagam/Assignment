## 2. Select Rows and Columns

In this section, we use `loc` and `iloc` to select specific rows and columns from the DataFrame.

We also use Boolean filtering to select rows based on a condition.

### Select Specific Columns

We can select one or more columns from a DataFrame by using the column names.

```python
df[["Name","Age","Revenue"]]
```

### Select Rows and Columns Using `loc`

The `loc` method selects data using row and column labels.

```python
df.loc[0:2,["Name","Age","Revenue"]]
```

### Select Rows and Columns Using `iloc`

The `iloc` method selects data using integer-based row and column positions.

```python
df.iloc[0:3,0:3]
```

### Filter Rows Using a Boolean Condition

Boolean filtering allows us to select only the rows that satisfy a condition.

Here, we select rows where `Age` is greater than 19.

```python
df[df["Age"]>19]
```

## 3. Create a Derived Column

A new column can be created from existing columns.

In this section, we create a `Revenue_per_Age` column by dividing `Revenue` by `Age`.

```python
df["Revenue_Age"]=df["Revenue"]/df["Age"]
df.head()
```

### Derived Column Result

The new `Revenue_Age` column is calculated using the existing `Revenue` and `Age` columns.

```python
df[["Name", "Revenue", "Age", "Revenue_Age"]]
```

## 4. Group By and Aggregate

In this section, we group the data by the categorical `Country` column.

We calculate the mean and count of the `Revenue` column for each country.

```python
df.groupby("Country")["Revenue"].agg(["mean","count"])
```
